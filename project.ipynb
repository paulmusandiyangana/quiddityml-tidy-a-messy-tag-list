## Step 1: The plan

**Goal:** find out how many different tags a survey collected, and print them as one clean line.

A survey asked people which topics they want to learn, and their answers were saved as one line of tags separated by commas. People typed the same tag in different ways: `NLP`, `nlp ` and ` NLP` all name the same topic. Python compares strings character by character, so it treats each spelling as a different tag.

You will first count the tags the way the raw line has them. Then you clean the line with string methods, drop the repeats with a set and sort what is left. The last step prints one summary line that starts with the number of tags, such as `3 unique tags: art, math, music`.

**The steps**

## Step 2: Look at the raw tags

**Goal:** see the raw survey line and how many tags it seems to hold before any cleaning.

`raw_tags` is one string with each answer separated by a comma. The cell prints it between square brackets, so you can see the spaces at the start and the newline at the end (the closing bracket sits on the next line). Then it splits the line at the commas without cleaning anything and counts the pieces and the different values among them. Both counts come out as 14, since ` NLP` and `  NLP` are different strings to Python.

```python
print(f"[{raw_tags}]")
pieces = raw_tags.split(",")
print(len(pieces), "pieces")
print(len(set(pieces)), "different values")
```

## Step 3: Clean the text and split it into tags

**Goal:** a list where each tag is spelled one way, so that the same topic gives the same string each time.

Start from `raw_tags` (step 2). Make `clean`: the line with the spaces and the newline taken off both ends, each letter lowercased and the spaces next to the commas removed. Then make `tags`: `clean` cut at each comma into a list of strings.

Each string method returns a new string and leaves the old one unchanged, so keep each result or chain the calls left to right. The repeats are still in the list, so `tags` has 14 items, one per answer.

```python
# TODO: strip the ends, lowercase, and remove the spaces next to the commas
clean = raw_tags.strip().lower().replace(' ', '')
# TODO: cut clean at each comma
tags = clean.split(",")

print(clean)
print(tags)
print(len(tags), "tags")
```

## Step 4: Drop the repeats and sort

**Goal:** each topic once, in a fixed order, so the list reads the same on each run.

Make `unique`: a set built from `tags` (step 3). A set keeps one copy of each value, so `nlp` is stored once however many times it appears in `tags`. A set has no order of its own. Make `ordered`: a new list with the items of `unique` in alphabetical order. The cell prints how many tags are left.

```python
# TODO: a set built from tags, then a sorted list built from that set
unique = set(tags)
ordered = sorted(unique)

print(len(unique), "unique tags")
print(ordered)
```

## Step 5: Print the summary line

**Goal:** one line a person can read at a glance: how many topics the survey collected and which ones.

Make `summary` with an f-string. It holds the number of tags in `ordered` (step 4) and the words `unique tags:`, followed by the tags with a comma and a space between them. For the tags `art` and `math` the line is `2 unique tags: art, math`.

```python
# TODO: an f-string with the number of tags and the tags joined by ", "
summary = f'{len(ordered)} unique tags: {', '.join(ordered)}'

print(summary)
```

## What I found

The tags looked different because of the inconsistent spacing, letter cases used. I used the strip() method to remove any whitespaces before and after each letter, lower() to turn all the tags into lower cases. The replace() method was used to remove spaces after the commas .The sort method was used to arrange the tags in alphabetical order and improve readability.
